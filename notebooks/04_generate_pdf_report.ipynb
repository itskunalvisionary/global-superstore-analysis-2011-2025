# 📄 Notebook 04 — Generate Final PDF Report

This notebook compiles all Matplotlib visuals generated in Notebook 03 into a professional PDF report.

## Includes:
- Yearly Sales Trend  
- Region-wise Sales Subplots  
- Category Sales Bar Chart  
- Sales Distribution Histogram  
- Discount vs Sales Scatter  
- Moving Average Trend  

Output file:
📁 reports/retail_report.pdf


```python
import matplotlib.pyplot as plt
from matplotlib.backends.backend_pdf import PdfPages
import os

# Create reports folder if missing
os.makedirs("reports", exist_ok=True)
```

```python
visuals = [
    "visuals/yearly_sales_trend.png",
    "visuals/region_sales_subplots.png",
    "visuals/category_sales_bar.png",
    "visuals/sales_histogram.png",
    "visuals/discount_vs_sales_scatter.png",
    "visuals/moving_average_trend.png"
]

visuals
```

```python
pdf_path = "reports/retail_report.pdf"

with PdfPages(pdf_path) as pdf:
    
    for img_path in visuals:
        # Read image
        img = plt.imread(img_path)
        
        # Create a figure for the PDF page
        plt.figure(figsize=(10, 6))
        plt.imshow(img)
        plt.axis("off")
        
        # Save the figure to the PDF
        pdf.savefig()
        plt.close()

pdf_path
```

```python
from matplotlib.backends.backend_pdf import PdfPages
import matplotlib.pyplot as plt

title_pdf_path = "reports/retail_report.pdf"

with PdfPages(title_pdf_path) as pdf:
    
    # Title page
    plt.figure(figsize=(10, 6))
    plt.text(0.5, 0.7, "Global SuperStore Sales Report", ha='center', fontsize=28, weight='bold')
    plt.text(0.5, 0.55, "Analysis Period: 2011–2024 & 2015-2024 AI Generated Data ", ha='center', fontsize=16)
    plt.text(0.5, 0.40, "Visualizations by Kunal Pandit", ha='center', fontsize=14, style='italic')
    plt.axis("off")
    pdf.savefig()
    plt.close()

    # Add each actual visualization
    for img_path in visuals:
        img = plt.imread(img_path)
        plt.figure(figsize=(10, 6))
        plt.imshow(img)
        plt.axis("off")
        pdf.savefig()
        plt.close()
```
