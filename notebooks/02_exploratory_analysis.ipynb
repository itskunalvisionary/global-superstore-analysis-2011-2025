# **Load Cleaned Data**

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

pd.set_option('display.max_columns', None)
```

**Import cleaned file**

```python
path = r"C:\Users\KUNAL\Downloads\AI ENGINEER\ProjectsX\global-superstore-analysis\data\superstore_cleaned.csv"
df = pd.read_csv(path,parse_dates=['order_date'])
```

```python
df.head(1)
```

```python
df.info()
df.describe(include='all')
```

**MISSING VALUE CHECK**

```python
df.isna().sum()
```

```python
plt.figure(figsize=(8,4))
sns.boxplot(x=df['sales'])
plt.title("Sales Outlier Check")
plt.show()
```

**TIME FEATURES**

```python
df["month"] = df["order_date"].dt.month
df["quarter"] = df["order_date"].dt.quarter

df.head()
```

```python
yearly = df.groupby("year")["sales"].sum().reset_index()
yearly
```

```python
monthly = df.groupby("month")["sales"].sum().reset_index()
monthly
```

```python
category = df.groupby("category")["sales"].sum().reset_index().sort_values('sales', ascending=False)
category
```

```python
region = df.groupby(["region","year",])["sales"].sum().reset_index()
region
```

```python
df.dtypes
```

```python
corr = df[["discount","profit","quantity","sales","shipping_cost"]].corr()

plt.figure(figsize=(8,6))
sns.heatmap(corr,annot=True,cmap="coolwarm")
plt.title("Correlation Heatmap")
plt.show()
```

```python
df = df.sort_values("order_date")
df["MA_3"] = df["sales"].rolling(3).mean()
df[["order_date","sales","MA_3"]].head()
```

```python
yearly.to_csv(r"C:\Users\KUNAL\Downloads\AI ENGINEER\ProjectsX\global-superstore-analysis\data\yearly_sales.csv", index=False)
monthly.to_csv(r"C:\Users\KUNAL\Downloads\AI ENGINEER\ProjectsX\global-superstore-analysis\data\monthly_sales.csv", index=False)
region.to_csv(r"C:\Users\KUNAL\Downloads\AI ENGINEER\ProjectsX\global-superstore-analysis\data\region_sales.csv", index=False)
category.to_csv(r"C:\Users\KUNAL\Downloads\AI ENGINEER\ProjectsX\global-superstore-analysis\data\category_sales.csv", index=False)
```
