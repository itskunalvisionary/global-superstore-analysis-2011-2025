# **DATA CLEANING PIPELINE**

```python
df.head(1)
```

**Change Columns Name**

```python
df.columns = [
    "category", "city", "country", "customer_id", "customer_name",
    "discount", "market", "record_count", "order_date", "order_id",
    "order_priority", "product_id", "product_name", "profit", "quantity",
    "region", "row_id", "sales", "segment", "ship_date", "ship_mode",
    "shipping_cost", "state", "sub_category", "year", "market2", "weeknum"
]
```

```python
df.head(1)
```

```python
df.dtypes
```

**Convert Dates**

```python
df['order_date'] = pd.to_datetime(df['order_date'], format='mixed')
df['ship_date'] = pd.to_datetime(df['ship_date'], format='mixed')
```

```python
df["order_date"].dtype
```

**Remove Duplicates**

```python
df = df.drop_duplicates()
```

**Clean Numeric Columns**

```python
num_cols = ['discount', 'record_count', 'profit', 'quantity', 'row_id', 'sales','shipping_cost', 'year', 'weeknum']
df[num_cols] = df[num_cols].apply(pd.to_numeric, errors="coerce")
```

**Handle Missing Values**

```python
df.head()
```

```python
df = df.dropna(subset=['order_id','customer_id','product_id','order_date','sales','profit','quantity'])
df["discount"] = df['discount'].fillna(df["discount"].mean())
df["shipping_cost"] = df["shipping_cost"].fillna(df['shipping_cost'].median())
df["segment"] = df['segment'].fillna(df['segment'].fillna("Unknown"))
```

**Save Cleaned Data**

```python
df.to_csv(r"C:\Users\KUNAL\Downloads\AI ENGINEER\ProjectsX\global-superstore-analysis\data\superstore_cleaned.csv",index=False)
```
